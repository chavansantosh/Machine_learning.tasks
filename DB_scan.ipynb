# import the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 

```python
# Load the dataset
df = pd.read_csv('dbscan_similar_2d_dataset.csv')
```

```python
df.head()
```

```python
# scatter plot
plt.scatter (df['x'], df['y'])
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot of Data')
plt.show()
```

```python
# import the model

from sklearn.cluster import DBSCAN

# create instance model
model = DBSCAN( eps=2,min_samples=10)

# train the model 
model.fit(df)

# display the labels
label = model.labels_
```

```python
# make the plot using the labels
plt.scatter(df["x"],df["y"],c = label)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('DBSCAN Clustering')
plt.show()
```
